```python
from google import genai

# Initialize the client
# client = genai.Client(api_key="YOUR_API_KEY")

def clear_all_file_search_stores():
    print("--- Deleting All File Search Stores ---")
    try:
        # List all stores
        stores = list(client.file_search_stores.list())
        
        if not stores:
            print("No File Search Stores found.")
            return

        for store in stores:
            print(f"\nProcessing store: {store.name}...")
            
            # STEP 1: Empty the store
            try:
                # List files specifically associated with this store
                # The 'parent' argument is the store name (e.g., fileSearchStores/xyz)
                store_files = list(client.file_search_stores.documents.list(parent=store.name))
                
                if store_files:
                    print(f"  - Found {len(store_files)} files in store. Removing associations...")
                    for store_file in store_files:
                        try:
                            # Delete the association between the file and the store
                            client.file_search_stores.documents.delete(name=store_file.name)
                        except Exception as e:
                            print(f"  - Failed to remove file {store_file.name}: {e}")
                else:
                    print("  - Store is already empty.")
                    
            except Exception as e:
                print(f"  - Error checking files in store: {e}")

            # STEP 2: Delete the store
            print(f"  - Deleting store container: {store.name}...")
            try:
                client.file_search_stores.delete(name=store.name)
                print(f"  - Successfully deleted {store.name}")
            except Exception as e:
                print(f"  - Failed to delete {store.name}: {e}")
                
    except Exception as e:
        print(f"Error listing stores: {e}")

def clear_all_uploaded_files():
    print("\n--- Deleting All Uploaded Files ---")
    try:
        # List all raw files
        files = list(client.files.list())
        
        if not files:
            print("No uploaded files found.")
            return

        for file in files:
            print(f"Deleting file: {file.name} ({file.display_name})...")
            try:
                client.files.delete(name=file.name)
                print(f"Successfully deleted {file.name}")
            except Exception as e:
                print(f"Failed to delete {file.name}: {e}")
    except Exception as e:
        print(f"Error listing files: {e}")

if __name__ == "__main__":
    # Step 1: Empty and Delete Stores (indexes)
    clear_all_file_search_stores()
    
    # Step 2: Delete Raw Files (source documents)
    clear_all_uploaded_files()
    
    print("\n--- Cleanup Complete ---")
```
